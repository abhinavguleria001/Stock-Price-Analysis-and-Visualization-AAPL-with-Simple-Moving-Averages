```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
import yfinance as yf
stock = "AAPL"
data = yf.download(stock,start = "2020-01-01", end = "2025-01-01")
print(data.head())
print(data.tail())

# Drop any rows with missing values
data = data.dropna()

# Check for null values
print(data.isnull().sum())

# Display summary statistics
print(data.describe())

# Calculate a 20-day Simple Moving Average
data['SMA_20'] = data['Close'].rolling(window=20).mean()

# Calculate a 50-day Simple Moving Average
data['SMA_50'] = data['Close'].rolling(window=50).mean()

# Calculate a 200-day Simple Moving Average
data['SMA_200'] = data['Close'].rolling(window=200).mean()

# Plot the closing price and the SMAs
plt.figure(figsize=(16, 8))  
plt.plot(data['Close'], label='Closing Price', linewidth=2) 
plt.plot(data['SMA_20'], label='20-day SMA', linestyle='--')  
plt.plot(data['SMA_50'], label='50-day SMA', linestyle='--')  
plt.plot(data['SMA_200'], label='200-day SMA', linestyle='--')  

plt.xlabel('Date', fontsize=14) 
plt.ylabel('Price', fontsize=14)  
plt.title(f'{stock} Closing Price with Simple Moving Averages', fontsize=16)
plt.legend(fontsize=12) 
plt.grid(True)

# Show the plot
plt.show()

# Print the last few rows of the DataFrame with the SMAs
print(data.tail())
```
